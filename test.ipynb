```python
import textarena as ta

# me = random.randint(0, 1)
me = 99

# Initialize agents
agents = {
    0: ta.agents.OpenRouterAgent(model_name="meta-llama/llama-4-maverick"),
    1: ta.agents.OpenRouterAgent(model_name="google/gemini-2.5-pro"),
}

# Initialize environment from subset and wrap it
env = ta.make(env_id="SimpleTak-v0-train")
dataset = []

env.reset(num_players=len(agents))
done = False
while not done:
    player_id, observation = env.get_observation()
    dataset.append({"id": len(final_dataset), "turn": len(dataset), "env_id": "SimpleTak-v0-train", "pid": player_id, "observation": observation})
    print(f'ADDED {len(dataset)} to the dataset')
    print(observation)
    if player_id == me or me == 3:
        action = input("Enter action: ")
    else:
        action = agents[player_id](observation)
    done, info = env.step(action=action)
rewards = env.close()
```
